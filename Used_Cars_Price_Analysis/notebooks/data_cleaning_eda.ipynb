### AGRA

```python
#loading data
agra_df=pd.read_csv(r"E:\Used_Cars_Analysis\data\raw\cars24_agra_raw.csv")
agra_df
```

```python
#first 5 rows
agra_df.head()
```

```python
#last 5 rows
agra_df.tail()
```

```python
#size of dataset
agra_df.shape
```

```python
#summary of the dataset
agra_df.info()
```

```python
#statistical summary of the dataset
agra_df.describe(include="all")
```

```python
#finding duplicates
agra_df.duplicated().sum()
```

```python
#finding null values
agra_df.isna().sum()
```

```python
#column names
agra_df.columns
```

```python
#dropping inessential columns
agra_df.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
agra_df.columns
```

```python
#renaming column
agra_df.rename(columns={'Distance Driven':'KMS_Driven'},inplace=True)
```

```python
#standardizing the column names
agra_df.columns = (agra_df.columns.str.lower().str.replace(" ","_"))
```

```python
agra_df.columns
```

```python
#datatypes of the data
agra_df.dtypes
```

```python
#standarding the data
agra_df['kms_driven'] = (agra_df['kms_driven'].str.replace(" km", "").str.replace(",", "").astype(int))
```

```python
agra_df['emi']=(agra_df['emi'].str.replace("EMI ₹","").str.replace(",","").str.replace("/m*","").astype(int))
```

```python
agra_df['price_num'] = agra_df['price'].str.extract(r'([\d.]+)').astype(float)

agra_df['is_crore'] = agra_df['price'].str.contains(' Crore', case=False, regex=True)

agra_df['price_lakhs'] = agra_df['price_num']
agra_df.loc[agra_df['is_crore'], 'price_lakhs'] = agra_df['price_num'] * 100
```

```python
agra_df.drop(['price','price_num','is_crore'],axis=1,inplace=True)
```

```python
agra_df.rename(columns={'price_lakhs':'price'},inplace=True)
```

```python
agra_df['price']=(agra_df['price']*100000)
```

```python
#current year to get the age of car
current_year=2026
```

```python
#feature engineering
#adding car_age_years column
agra_df['car_age_years']=current_year-agra_df['year_of_manufacture']
```

```python
#adding city column
agra_df['city'] = 'Agra'
```

```python
agra_df.head()
```

```python
#outliers detection ( IQR Method )
num_cols = ['kms_driven','emi','price']
for col in num_cols:
    Q1 = agra_df[col].quantile(0.25)
    Q3 = agra_df[col].quantile(0.75)
    IQR = Q3-Q1
    lower = Q1-1.5*IQR
    upper = Q3+1.5*IQR
    agra_outliers = agra_df[(agra_df[col]<= lower) | (agra_df[col]>= upper)]
agra_outliers

#As these outliers are real and normal , they are not errors so we are not dropping them
```

```python
#final validation
agra_df.isnull().sum()
```

```python
agra_df.to_csv("cars24_agra_cleaned.csv")
```
